```python
import os
import re
```

```python
directory = "files"

filenames = os.listdir(directory)
print(filenames)
```

```python


for filename in filenames:
    filepath = os.path.join(directory, filename)
    with open(filepath, "r") as file:
        content = file.read()
        pattern = r"[a-zA-Z0-9 ,'()]+"
        text = re.findall(pattern, content)
        print(text[0])
```
