```python
import requests
from bs4 import BeautifulSoup
```

```python
enter_lat = 30.0444
enter_lon = 31.2357
```

```python
url = f"https://weather.com/weather/today/l/{enter_lat},{enter_lon}"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

city = soup.find("h1", class_="CurrentConditions--location--yub4l").get_text()
temp = soup.find("span", class_="CurrentConditions--tempValue--zUBSz").get_text()
feels_like = soup.find("span", class_="TodayDetailsCard--feelsLikeTempValue--8WgHV").get_text()
print(f"Enter the latitude: {enter_lat}")
print(f"Enter the Longitude: {enter_lon}")
print(f"Current Location: {city}")
print(f"Current Temperature: {temp[:len(temp) - 1]} F")
print(f"Current Temperature: {temp} F and {round((int(temp[:len(temp) - 1])-32)*(5/9))} C")
print(f"Feels Like: {feels_like} F and {round((int(feels_like[:len(feels_like) - 1])-32)*(5/9))} C")
```
