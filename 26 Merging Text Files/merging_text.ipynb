```python
import os
import glob
```

```python
filepaths = glob.glob("files_text/*.txt")
print(filepaths)
```

```python
content = ""

for filepath in filepaths:
    with open(filepath, "r", encoding="utf-8") as file:
        content += file.read()
        content += " "

print(content)
```

```python
with open("output.txt", "w") as file:
    file.write(content)
    print(f"File successfully saved !!!")
```
